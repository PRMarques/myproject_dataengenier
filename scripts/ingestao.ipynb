```python
import duckdb
import os
import sys
import pandas as pd
from datetime import datetime
```

```python
con = duckdb.connect(database='dados_duckdb.db', read_only=False)
```

```python
arquivo = 'z0019_2.csv' ## nome do arquivo a ser lido
data_ingestao = datetime.now() ## pega a data e hora atual
df = pd.read_csv(f'../landing/{arquivo}', sep=';') ## lê o arquivo csv
df['nome_arquivo'] = arquivo ## adiciona o nome do arquivo
df['data_ingestao'] = data_ingestao ## adiciona a data de ingestão
df.head() ## mostra as 5 primeiras linhas do dataframe
```

```python
con.execute("""CREATE TABLE IF NOT EXISTS bronze_produtos(
    NATBR VARCHAR(255),
    MAKTX_produto VARCHAR(255),
    WERKS VARCHAR(255),
    AINS DECIMAL(10,2),
    LABST VARCHAR(255),
    nome_arquivo VARCHAR(255),
    data_ingestao TIMESTAMP
    )
""")## cria uma tabela bronze_produtos no banco de dados DuckDB com os dados do dataframe df
```

```python
con.execute("""INSERT INTO bronze_produtos SELECT * FROM df""") ## insere os dados do dataframe df na tabela bronze_produtos
```

```python
resultado = con.execute("""SELECT * FROM bronze_z0019""").fetchdf() ## seleciona todos os dados da tabela bronze_produtos e armazena em um dataframe resultado
resultado.head(6) ## mostra as 5 primeiras linhas do dataframe resultado
```

```python
con.execute(""" ALTER TABLE bronze_produtos RENAME TO bronze_z0019""") ## renomeia a tabela bronze_produtos para bronze_z0019
```
